Q2. WAP to encrypt and decrypt the message "Hello World" using SHA256.




```python
import hashlib

def encrypt_message(message):
    # Hash the message using SHA-256
    sha_signature = hashlib.sha256(message.encode()).hexdigest()
    return sha_signature

def verify_message(message, hashed_value):
    # Re-hash the message and compare with stored hash
    return encrypt_message(message) == hashed_value

# Example usage
message = "Hello World"
encrypted = encrypt_message(message)
print("Encrypted (SHA-256 Hash):", encrypted)

# Verification (simulating 'decryption')
is_valid = verify_message("Hello World", encrypted)
print("Message verified:", is_valid)

```

Q4. Create a simple blockchain using proof of work (PoW)

```python
import hashlib
import time

class Block:
    def __init__(self, index, timestamp, data, previous_hash, nonce=0):
        self.index = index
        self.timestamp = timestamp
        self.data = data
        self.previous_hash = previous_hash
        self.nonce = nonce
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        sha = hashlib.sha256()
        sha.update(str(self.index).encode('utf-8') +
                   str(self.timestamp).encode('utf-8') +
                   str(self.data).encode('utf-8') +
                   str(self.previous_hash).encode('utf-8') +
                   str(self.nonce).encode('utf-8'))
        return sha.hexdigest()

    def mine_block(self, difficulty):
        while self.hash[:difficulty] != '0' * difficulty:
            self.nonce += 1
            self.hash = self.calculate_hash()
        print("Block Mined! Hash: " + self.hash)

class Blockchain:
    def __init__(self):
        self.chain = [self.create_genesis_block()]
        self.difficulty = 4

    def create_genesis_block(self):
        return Block(0, time.time(), "Genesis Block", "0")

    def get_latest_block(self):
        return self.chain[-1]

    def add_block(self, new_block):
        new_block.previous_hash = self.get_latest_block().hash
        new_block.mine_block(self.difficulty)
        self.chain.append(new_block)

    def is_chain_valid(self):
        for i in range(1, len(self.chain)):
            current_block = self.chain[i]
            previous_block = self.chain[i-1]
            if current_block.hash != current_block.calculate_hash():
                return False
            if current_block.previous_hash != previous_block.hash:
                return False
        return True

# Execution
my_blockchain = Blockchain()
print("Mining block 1...")
my_blockchain.add_block(Block(1, time.time(), {"amount": 10}, ""))

print("Mining block 2...")
my_blockchain.add_block(Block(2, time.time(), {"amount": 20}, ""))

print("Blockchain valid:", my_blockchain.is_chain_valid())
```

Q5. Demonstrate sending of a digitally signed document.

```python
!pip install pycryptodome
from Crypto.PublicKey import RSA
from Crypto.Signature import pkcs1_15
from Crypto.Hash import SHA256

# 1. Generate Keys (Sender's keys)
key = RSA.generate(2048)
private_key = key
public_key = key.publickey()

# 2. The Document/Message to be 'sent'
document = b"This is a confidential document content."

# 3. Signing the document (Sender side)
hash_obj = SHA256.new(document)
signature = pkcs1_15.new(private_key).sign(hash_obj)

print("--- Sender Side ---")
print(f"Document: {document.decode()}")
print(f"Signature generated (hex): {signature.hex()[:50]}...")

# 4. Verifying the document (Receiver side)
received_document = document
received_signature = signature

try:
    hash_to_verify = SHA256.new(received_document)
    pkcs1_15.new(public_key).verify(hash_to_verify, received_signature)
    print("\n--- Receiver Side ---")
    print("Verification Result: Signature is VALID. The document is authentic.")
except (ValueError, TypeError):
    print("\n--- Receiver Side ---")
    print("Verification Result: Signature is INVALID. The document may have been altered.")

# Simulating Tampering
print("\n--- Tampering Scenario ---")
tampered_document = b"This is a modified document content."
try:
    hash_tampered = SHA256.new(tampered_document)
    pkcs1_15.new(public_key).verify(hash_tampered, received_signature)
except (ValueError, TypeError):
    print("Verification Result for tampered document: Signature is INVALID.")
```

Q6. Create a BlockChain block containing block hash, transaction history, time of creation

```python
import hashlib
import time
import json

class Block:
    def __init__(self, index, transactions, previous_hash):
        self.index = index
        self.timestamp = time.ctime() # Time of creation
        self.transactions = transactions # Transaction history
        self.previous_hash = previous_hash
        self.nonce = 0
        self.hash = self.calculate_hash() # Block hash

    def calculate_hash(self):
        block_string = json.dumps({
            "index": self.index,
            "timestamp": self.timestamp,
            "transactions": self.transactions,
            "previous_hash": self.previous_hash,
            "nonce": self.nonce
        }, sort_keys=True).encode()
        return hashlib.sha256(block_string).hexdigest()

# Example of creating a block
transaction_history = [
    {"from": "Alice", "to": "Bob", "amount": 50},
    {"from": "Bob", "to": "Charlie", "amount": 25}
]

new_block = Block(index=1, transactions=transaction_history, previous_hash="0000abc123")

print(f"Block Index: {new_block.index}")
print(f"Time of Creation: {new_block.timestamp}")
print(f"Transaction History: {new_block.transactions}")
print(f"Previous Hash: {new_block.previous_hash}")
print(f"Block Hash: {new_block.hash}")
```

Q8. Create a blockchain of 5 nodes and check its validity

```python
import hashlib
import time

class Block:
    def __init__(self, index, timestamp, data, previous_hash, nonce=0):
        self.index = index
        self.timestamp = timestamp
        self.data = data
        self.previous_hash = previous_hash
        self.nonce = nonce
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        sha = hashlib.sha256()
        sha.update(str(self.index).encode('utf-8') +
                   str(self.timestamp).encode('utf-8') +
                   str(self.data).encode('utf-8') +
                   str(self.previous_hash).encode('utf-8') +
                   str(self.nonce).encode('utf-8'))
        return sha.hexdigest()

    def mine_block(self, difficulty):
        while self.hash[:difficulty] != '0' * difficulty:
            self.nonce += 1
            self.hash = self.calculate_hash()

class Blockchain:
    def __init__(self):
        self.chain = [self.create_genesis_block()]
        self.difficulty = 2

    def create_genesis_block(self):
        return Block(0, time.time(), "Genesis Block", "0")

    def get_latest_block(self):
        return self.chain[-1]

    def add_block(self, new_block):
        new_block.previous_hash = self.get_latest_block().hash
        new_block.mine_block(self.difficulty)
        self.chain.append(new_block)

    def is_chain_valid(self):
        for i in range(1, len(self.chain)):
            current_block = self.chain[i]
            previous_block = self.chain[i-1]
            if current_block.hash != current_block.calculate_hash():
                print(f"Invalid hash at block {i}")
                return False
            if current_block.previous_hash != previous_block.hash:
                print(f"Hash link broken at block {i}")
                return False
        return True

# Execution
bc = Blockchain()
for i in range(1, 5):
    print(f"Mining block {i+1}...")
    bc.add_block(Block(i, time.time(), f"Data for block {i+1}", ""))

print("\n--- Blockchain Status ---")
for block in bc.chain:
    print(f"Block {block.index} Hash: {block.hash}")

print("\nIs Blockchain Valid?", bc.is_chain_valid())
```

Q10. Create a simple permissioned blockchain using Hyperledger fabric.

```python
import hashlib
import time

class PermissionedBlock:
    def __init__(self, index, data, previous_hash, creator_identity):
        self.index = index
        self.timestamp = time.ctime()
        self.data = data
        self.previous_hash = previous_hash
        self.creator = creator_identity
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        block_string = f"{self.index}{self.timestamp}{self.data}{self.previous_hash}{self.creator}"
        return hashlib.sha256(block_string.encode()).hexdigest()

class PermissionedBlockchain:
    def __init__(self):
        self.chain = [self.create_genesis_block()]
        # Access Control List (Authorized identities)
        self.authorized_nodes = {"Admin_Node", "Validator_1", "Validator_2"}

    def create_genesis_block(self):
        return PermissionedBlock(0, "Genesis Block", "0", "System")

    def add_block(self, data, creator_identity):
        # Permission Check (Simulating Hyperledger MSP logic)
        if creator_identity not in self.authorized_nodes:
            print(f"Access Denied: Node '{creator_identity}' is not authorized to write to this ledger.")
            return False

        previous_block = self.chain[-1]
        new_block = PermissionedBlock(len(self.chain), data, previous_block.hash, creator_identity)
        self.chain.append(new_block)
        print(f"Block {new_block.index} added successfully by {creator_identity}.")
        return True

# Execution
fabric_sim = PermissionedBlockchain()

print("--- Attempting transactions ---")
# Authorized transaction
fabric_sim.add_block("Transaction: Alice pays Bob $10", "Validator_1")

# Unauthorized transaction (simulating a malicious or unknown node)
fabric_sim.add_block("Transaction: Malicious activity", "Guest_Node")

# Authorized transaction
fabric_sim.add_block("Transaction: Bob pays Charlie $5", "Admin_Node")

print("\n--- Ledger State ---")
for block in fabric_sim.chain:
    print(f"Index: {block.index} | Creator: {block.creator} | Hash: {block.hash[:20]}...")
```
